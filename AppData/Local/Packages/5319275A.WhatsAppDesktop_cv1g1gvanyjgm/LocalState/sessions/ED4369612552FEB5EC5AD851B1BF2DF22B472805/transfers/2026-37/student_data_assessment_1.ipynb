## Section 5: Data Analysis & Visualization

### Introduction to Different Types of Analysis

Now that your data is clean, it's time to ask questions and find answers. There are three ways to look at data:

#### 1. **Univariate Analysis** (One Variable)

You look at **one variable in isolation** to understand its distribution and characteristics.

**In plain English:** "What does this one thing look like by itself?"

**Examples:**
- How old are our customers? (one variable: Age)
- How much do people spend? (one variable: Purchase_Amount)
- What product categories do we sell? (one variable: Product_Category)

**What you want to know:**
- The range (minimum to maximum)
- Where most values cluster (are they bunched up, spread out, skewed to one side?)
- The frequency (how many, how often)

---

#### 2. **Bivariate Analysis** (Two Variables)

You look at **two variables together** to see if they're related or influence each other.

**In plain English:** "Does this thing change when that thing changes?"

**Examples:**
- Do older customers spend more money? (Age → Purchase_Amount)
- Do different genders prefer different product categories? (Gender → Product_Category)
- Do loyal customers (Gold status) spend more than casual customers (Bronze)? (Loyalty_Status → Purchase_Amount)

**What you want to know:**
- Is there a pattern or relationship?
- If one variable goes up, does the other go up too? (positive relationship)
- Or does one go up while the other goes down? (negative relationship)
- Or is there no relationship at all?

---

#### 3. **Multivariate Analysis** (Three or More Variables)

You look at **multiple variables together** to understand complex relationships and segments.

**In plain English:** "How do many things interact together to tell a bigger story?"

**Examples:**
- For each product category, what's the average spending by age group and gender?
- Which customer segments (by age + loyalty status) are most valuable (spend the most)?
- How does tenure + loyalty status + age affect spending?

**What you want to know:**
- The complete picture across multiple dimensions
- Customer segments and their characteristics
- Complex patterns that only show up when you look at many variables together

---

### Choosing the Right Visualization

Different questions need different chart types. Think about what you're trying to show:

| Type | Question | Best Visualization | What It Shows |
|---|---|---|---|
| **Univariate** | How are values spread across a range? | **Something that shows frequency across a range** | Distribution of continuous data |
| **Univariate** | What are the counts for each category? | **Something that compares counts side-by-side** | Frequency of categorical data |
| **Bivariate** | How do two numeric variables relate? | **Something that shows points with a relationship** | Correlation and trends |
| **Bivariate** | How does numeric data differ across groups? | **Something that shows distribution within each group** | Comparison of distributions |
| **Bivariate** | How do two categorical variables relate? | **Something that shows grouped counts** | Patterns across categories |
| **Multivariate** | How do 3+ variables interact in segments? | **Something that shows multiple dimensions at once** | Complex patterns and segments |
